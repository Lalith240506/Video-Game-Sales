import streamlit as st
import pandas as pd
import numpy as np
import plotly.express as px
import plotly.graph_objects as go
import os

```python
st.set_page_config(page_title="Video Game Sales Dashboard", layout="wide", initial_sidebar_state="expanded")
```

```python
NEON_COLORS = ['#00f3ff', '#ff00ff', '#39ff14', '#fdfa72', '#ff073a', '#bc13fe', '#ff71ce', '#fce83a']
BG_TRANSPARENT = 'rgba(0,0,0,0)'
```

```python
def apply_cyberpunk_theme(fig):
    fig.update_layout(
        template='plotly_dark',
        plot_bgcolor=BG_TRANSPARENT,
        paper_bgcolor=BG_TRANSPARENT,
        font=dict(color='#E0E0E0', family="sans-serif"),
        title=dict(font=dict(size=20, color='#00f3ff')),
        legend=dict(bgcolor='rgba(18, 21, 30, 0.5)'),
        margin=dict(l=40, r=40, t=60, b=40)
    )
    fig.update_xaxes(showgrid=True, gridwidth=1, gridcolor='rgba(255, 255, 255, 0.1)')
    fig.update_yaxes(showgrid=True, gridwidth=1, gridcolor='rgba(255, 255, 255, 0.1)')
    return fig
```

```python
@st.cache_data
def load_data():
    file_path = 'vgsales.csv'
    if os.path.exists(file_path):
        return pd.read_csv(file_path)
    
    np.random.seed(42)
    n = 2000
    genres = ['Action', 'RPG', 'Shooter', 'Sports', 'Platform', 'Racing', 'Strategy', 'Simulation', 'Fighting', 'Puzzle']
    publishers = ['Nintendo', 'Electronic Arts', 'Activision', 'Sony', 'Ubisoft', 'Take-Two', 'Sega', 'Square Enix', 'Capcom', 'Bandai Namco']
    platforms = ['PS4', 'PS3', 'X360', 'XOne', 'PC', 'Switch', 'Wii', 'DS', 'PS2', 'PS5', 'XSX']
    
    years = np.random.normal(loc=2010, scale=8, size=n).astype(int)
    years = np.clip(years, 1980, 2024)
    
    df = pd.DataFrame({
        'Name': [f"Game {i}" for i in range(n)],
        'Platform': np.random.choice(platforms, size=n),
        'Year': years,
        'Genre': np.random.choice(genres, size=n),
        'Publisher': np.random.choice(publishers, size=n),
        'NA_Sales': np.random.exponential(1.5, size=n),
        'EU_Sales': np.random.exponential(1.0, size=n),
        'JP_Sales': np.random.exponential(0.5, size=n),
        'Other_Sales': np.random.exponential(0.3, size=n),
        'Critic_Score': np.clip(np.random.normal(72, 14, size=n), 10, 100).astype(int)
    })
    df['Global_Sales'] = df[['NA_Sales', 'EU_Sales', 'JP_Sales', 'Other_Sales']].sum(axis=1)
    df = df.round(2)
    df.to_csv(file_path, index=False)
    return df
```

```python
st.markdown("<h1 style='text-align: center; color: #00f3ff; text-shadow: 0 0 10px #00f3ff;'>VIDEO GAME SALES DASHBOARD</h1>", unsafe_allow_html=True)
st.markdown("<hr style='border: 1px solid rgba(0, 243, 255, 0.3);'>", unsafe_allow_html=True)
```

```python
st.sidebar.markdown("<h2 style='color: #39ff14;'>CYBER-FILTERS</h2>", unsafe_allow_html=True)
selected_genres = st.sidebar.multiselect("SELECT GENRE", sorted(df['Genre'].unique()), default=df['Genre'].unique())
year_range = st.sidebar.slider("YEAR RANGE", int(df['Year'].min()), int(df['Year'].max()), (1990, 2024))

filtered_df = df[(df['Genre'].isin(selected_genres)) & (df['Year'].between(year_range[0], year_range[1]))]
```

```python
k1, k2, k3 = st.columns(3)
k1.metric("Global Sales", f"{filtered_df['Global_Sales'].sum():,.1f}M")
k2.metric("Avg Critic Score", f"{filtered_df['Critic_Score'].mean():.1f}")
k3.metric("Record Count", f"{len(filtered_df):,}")
```

```python
c1, c2 = st.columns([2, 1])
with c1:
    line_data = filtered_df.groupby('Year')['Global_Sales'].sum().reset_index()
    fig = px.line(line_data, x='Year', y='Global_Sales', title="Sales Trend", color_discrete_sequence=['#00f3ff'])
    st.plotly_chart(apply_cyberpunk_theme(fig), use_container_width=True)

with c2:
    pie_data = filtered_df.groupby('Genre')['Global_Sales'].sum().reset_index()
    fig = px.pie(pie_data, names='Genre', values='Global_Sales', hole=0.5, color_discrete_sequence=NEON_COLORS)
    st.plotly_chart(apply_cyberpunk_theme(fig), use_container_width=True)

st.dataframe(filtered_df.head(100), use_container_width=True)
```
